## 3. Conteo de viviendas y valores nulos

- **¿Cuántas viviendas hay en el conjunto de datos?**
- **Variables con valores nulos y recuento de nulos por variable.**

```python
# Número de viviendas en el conjunto de datos
num_viviendas = df.shape[0]
print(f"Número total de viviendas: {num_viviendas}")
```

```python
# Variables con valores nulos y su cantidad
nulos = df.isnull().sum()
variables_con_nulos = nulos[nulos > 0].sort_values(ascending=False)

print(f"Hay {len(variables_con_nulos)} variables con valores nulos:
")
print(variables_con_nulos)
```

```python
# Representación en un DataFrame con porcentaje de nulos
df_nulos = pd.DataFrame({
    'Valores Nulos': variables_con_nulos,
    'Porcentaje (%)': (variables_con_nulos / num_viviendas * 100).round(2)
})
df_nulos
```

> **Nota según :** En varias de estas columnas (como , , , , , variables de  y ), el valor  en realidad codifica la ausencia de dicho elemento (por ejemplo, "Sin piscina", "Sin garaje", "Sin sótano", "Sin chimenea"), y no necesariamente una pérdida de información accidental.
